# Ampliación opcional - El reto del analista
Mediante pandas, responde e interpreta:

1. ¿Qué nivel de fidelidad presenta mayor gasto mensual medio?
2. ¿El gasto histórico medio difiere entre quienes compran el mes siguiente y quienes no?
3. ¿Qué dispositivo tiene mayor número medio de compras?
4. ¿Cuál es la satisfacción media por nivel de fidelidad?
5. ¿Qué porcentaje de clientes de cada nivel compra el mes siguiente?

> Una asociación no demuestra causalidad.

````python
# --- Ampliación opcional ---

# 1) Nivel de fidelidad con mayor gasto mensual medio
print("1) Gasto mensual medio por nivel de fidelidad:")
print(df.groupby("nivel_fidelidad")["gasto_mensual"].mean().sort_values(ascending=False))
print()

# 2) Gasto medio según si compra o no el mes siguiente
print("2) Gasto mensual medio según compra_proximo_mes:")
print(df.groupby("compra_proximo_mes")["gasto_mensual"].mean())
print()

# 3) Dispositivo con mayor número medio de compras
print("3) Número medio de compras por dispositivo:")
print(df.groupby("dispositivo")["num_compras"].mean().sort_values(ascending=False))
print()

# 4) Satisfacción media por nivel de fidelidad
print("4) Satisfacción media por nivel de fidelidad:")
print(df.groupby("nivel_fidelidad")["satisfaccion"].mean().sort_values(ascending=False))
print()

# 5) Porcentaje de clientes de cada nivel que compra el mes siguiente
print("5) % de compra (Sí) por nivel de fidelidad:")
pct = pd.crosstab(df["nivel_fidelidad"], df["compra_proximo_mes"], normalize="index") * 100
print(pct.round(2))

# Interpretación breve:
# 1) Oro tiene el gasto medio más alto (~70,9 €).
# 2) Quienes compran el mes siguiente gastan de media más (~76 € vs ~65 €).
# 3) Ordenador y Tablet tienen medias de compras muy parecidas (~8,24).
# 4) Oro y Plata tienen la satisfacción media más alta (~3,37).
# 5) Platino es el que más compra el mes siguiente (~62%).
# Recuerda: asociación ≠ causalidad.
````
